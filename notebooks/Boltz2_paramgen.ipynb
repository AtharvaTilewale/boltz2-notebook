# Boltz2 Parameter Generator

Generate YAML configuration files (`params.yaml`) and runtime parameters for Boltz2 biomolecular structure prediction.

- Supports Protein, DNA, RNA, and Ligand (SMILES / CCD) entities
- Configure covalent bonds, pocket conditioning, and template guidance
- Save and download ready-to-run YAML parameters

```python
# @title Setup Parameter Generator Environment
import os, sys, subprocess, shutil

os.chdir("/content")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pyyaml"], check=False)

if not os.path.exists("/content/boltz2-notebook"):
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/AtharvaTilewale/boltz2-notebook.git"], check=True)

os.makedirs("/content/boltz_data/scripts/v2", exist_ok=True)
src_script = "/content/boltz2-notebook/scripts/v2/param_gen.py"
dst_script = "/content/boltz_data/scripts/v2/param_gen.py"
if os.path.exists(src_script):
    shutil.copy2(src_script, dst_script)

print("Parameter Generator environment ready!")
```

```python
# @title Launch Parameter Generator GUI
%run /content/boltz_data/scripts/v2/param_gen.py
```

```python
# @title Download Generated YAML
from google.colab import files
import os

param_file = "/content/boltz_data/params.yaml"
if os.path.exists(param_file):
    files.download(param_file)
    print(f"Downloaded {param_file}")
else:
    print("params.yaml not found. Please click 'Save YAML' in the generator above first.")
```
