# Boltz2 Parameter Generator

Generate YAML configuration files (`params.yaml`) and runtime parameters for Boltz2 biomolecular structure prediction.

- Supports Protein, DNA, RNA, and Ligand (SMILES / CCD) entities
- Configure covalent bonds, pocket conditioning, and template guidance
- Save and download ready-to-run YAML parameters
