# Boltz2 Parameter Generator

Generate YAML configuration files (`params.yaml`) and runtime parameters for Boltz2 biomolecular structure prediction.

- Supports Protein, DNA, RNA, and Ligand (SMILES / CCD) entities
- Configure covalent bonds, pocket conditioning, and template guidance
- Save and download ready-to-run YAML parameters

> ----------------**Important Note**----------------<br>
> <br>**Colab Runtime & Python Compatibility:**  
> Due to recent changes in Colab, Colab is using Python 3.13, which makes dependencies installation much slower and takes a lot of time, some are incompatible as well. We recommend to switch to runtime version 2026.7 before running.  
>  
> **How to switch runtime version & enable GPU:**  
> 1. In Colab top menu, go to **Runtime > Change runtime type**.  
> 2. Under **Hardware accelerator**, select **GPU** (T4 or higher).  
> 3. Under **Runtime version**, select **2026.7** (Python 3.12).  
> 4. Click **Save** and **Connect** (top-right) before running any cells.  
> ----------------------------------------------------


```python
# @title Setup Parameter Generator Environment
# @markdown Fetches the notebook code and installs the lightweight `boltz2nb` package (no Boltz engine, no GPU needed).
NOTEBOOK_VERSION = "v2.1.3"  # @param {type:"string"}
import os, shutil, subprocess, sys
os.chdir("/content")
REPO = "/content/boltz2-notebook"
os.environ["BOLTZ2NB_BASE_DIR"] = "/content/boltz_data"
def _clone(ref):
    return subprocess.run(["git", "clone", "--quiet", "--depth", "1", "--branch", ref,
                           "https://github.com/AtharvaTilewale/boltz2-notebook.git", REPO]).returncode == 0
if not os.path.exists(REPO):
    if not _clone(NOTEBOOK_VERSION):
        print(f"[!] '{NOTEBOOK_VERSION}' not found; using 'main'."); _clone("main")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", f"{REPO}[ligand]"], check=True)
os.makedirs("/content/boltz_data/scripts", exist_ok=True)
shutil.copytree(f"{REPO}/scripts", "/content/boltz_data/scripts", dirs_exist_ok=True)
print("Parameter Generator environment ready.")
```

```python
# @title Launch Parameter Generator GUI
%run /content/boltz_data/scripts/v2/param_gen.py
```

```python
# @title Download Generated YAML and run parameters
from google.colab import files
import os
for f in ["/content/boltz_data/params.yaml", "/content/boltz_data/run_params.txt"]:
    if os.path.exists(f):
        files.download(f); print("Downloaded", f)
    else:
        print(f"{os.path.basename(f)} not found - click the corresponding Save button in the generator first.")
```
